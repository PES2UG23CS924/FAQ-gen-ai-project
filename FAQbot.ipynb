```python
with open("SampleCompanyPolicy.txt", "r", encoding="utf-8") as file:
    context = file.read()
print("Policy document loaded successfully!")
```

```python
from transformers import pipeline
qa_bot = pipeline(
"question-answering",
model="distilbert-base-cased-distilled-squad"
)
```

```python
question = "How many days of paid leave are employees entitled to?"

result = qa_bot(question=question, context=context)

print("Question:", question)
print("Answer:", result["answer"])
print("Confidence Score:", round(result["score"], 2))
```

```python
q= "Who should salary discrepancies be reported to?"
res= qa_bot(question=q, context=context)
print("Question: ",q)
print("Answer: ",res["answer"])
print("Confidence Score:", round(res["score"], 2))
```

```python
while True:
    question = input("Ask your question: ")
    if question.lower() == "exit":
        print("Exiting FAQ Bot. Thank you!")
        break   
    result = qa_bot(question=question, context=context)
    print("\nAnswer:", result["answer"],)
    print("Confidence Score:", round(result["score"], 2),"\n")
    if result["score"] < 0.3:
        print("Answer not confidently found in the document.\n")
```
